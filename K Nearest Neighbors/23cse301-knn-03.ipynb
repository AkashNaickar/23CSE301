# Exercise 2: Euclidean Distance Calculation

Calculate the distance between a new entry and other existing values using the Euclidean distance formula.

```python
import pandas as pd
import numpy as np
```

## Create Dataset

```python
data = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
})
print(data)
```

## Define Euclidean Distance Function

```python
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((point1 - point2) ** 2))
```

## Calculate Distance from New Entry to All Existing Values

```python
new_entry = np.array([45, 35])  # Brightness=45, Saturation=35

distances = []
for idx, row in data.iterrows():
    existing_point = np.array([row['Brightness'], row['Saturation']])
    dist = euclidean_distance(new_entry, existing_point)
    distances.append(dist)
    
data['Distance from (45,35)'] = [round(d, 2) for d in distances]
print(data.to_string(index=False))
```

## Classify by Nearest Neighbor (k=1)

```python
min_idx = data['Distance from (45,35)'].idxmin()
print(f"Nearest neighbor is index {min_idx}")
print(f"  Brightness={data.loc[min_idx, 'Brightness']}, Saturation={data.loc[min_idx, 'Saturation']}")
print(f"  Distance = {data.loc[min_idx, 'Distance from (45,35)']}")
print(f"  Class = {data.loc[min_idx, 'Class']}")
print(f"\nPrediction: The new entry (45, 35) is classified as '{data.loc[min_idx, 'Class']}'")
```
