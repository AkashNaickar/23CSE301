# 11.3 Implementing PCA using Python

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from mpl_toolkits.mplot3d import Axes3D
```

## Load Breast Cancer Dataset

```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
data.keys()
```

## Inspect Target Names and Feature Names

```python
print(data['target_names'])
print(data['feature_names'])
```

## Create DataFrame

```python
dfl1 = pd.DataFrame(data['data'], columns=data['feature_names'])
print(dfl1.shape)
dfl1.head()
```

## Standard Scaler and PCA

```python
sc = StandardScaler()
sc.fit(dfl1)
scaled_data = sc.transform(dfl1)

principal = PCA(n_components=3)
principal.fit(scaled_data)
x = principal.transform(scaled_data)
print(x.shape)
```

## Components

```python
print(principal.components_)
```

## 2D PCA Scatter Plot

```python
plt.figure(figsize=(10, 10))
plt.title('PCA 2D')
plt.scatter(x[:, 0], x[:, 1], c=data['target'], cmap='plasma')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Target')
plt.show()
```

## 3D PCA Scatter Plot

```python
fig = plt.figure(figsize=(10, 10))
axis = fig.add_subplot(111, projection='3d')
axis.scatter(x[:, 0], x[:, 1], x[:, 2], c=data['target'], cmap='plasma')
axis.set_xlabel('PC1', fontsize=10)
axis.set_ylabel('PC2', fontsize=10)
axis.set_zlabel('PC3', fontsize=10)
plt.title('PCA 3D')
plt.show()
```

## Explained Variance Ratio

```python
print(principal.explained_variance_ratio_)
```
